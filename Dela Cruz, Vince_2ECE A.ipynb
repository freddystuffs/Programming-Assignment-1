### A. WORD ROTATION PROBLEM

```python
def rotate_word(x):
    txt = x[1::]
    return(txt + x[0])
```

```python
rotate_word("python")
```

```python
rotate_word("logic")
```

```python
rotate_word("Code")
```

```python
rotate_word("A")
```

### B. USERNAME BUILDER PROBLEM

```python
def make_username(x,y):
    txt1 = x.lower()
    txt2 = y.lower()
    ftxt1 = txt1.replace(" ", "")
    ftxt2 = txt2.replace(" ", "")
    ftxt = ftxt1 + "." + ftxt2
    return(ftxt)
```

```python
make_username("Ada", "Lovelace")
```

```python
make_username("Alan", "Turing")
```

```python
make_username("Ana Maria", "De Leon")
```

### C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(items):
    first, *middle, last = items
    return[last, *middle, first]
```

```python
swap_bookends([1, 2, 3, 4, 5, 6])
```

```python
swap_bookends(["red", "green", "blue"])
```

```python
swap_bookends([8, 3]) 
```
